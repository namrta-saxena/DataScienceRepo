### Zero-Shot Text Classification

Using the `zero-shot-classification` pipeline from the `transformers` library, we can classify the `test_array` against the `candidate_labels` without any prior training. This is a powerful technique for categorizing text based on custom labels.

```python
from transformers import pipeline

# Load the zero-shot classification pipeline
# We'll use a general-purpose model like 'facebook/bart-large-mnli'
classifier = pipeline('zero-shot-classification', model='facebook/bart-large-mnli')

# Perform zero-shot classification
# The results will show scores for each candidate label for every text in test_array
zero_shot_results = classifier(test_array, candidate_labels=candidate_labels)

# Display the results in a more readable format
for i, item in enumerate(test_array):
    print(f"Original Text: '{item}'")
    print("Classification Results:")
    # Sort labels by score in descending order for better readability
    sorted_labels_scores = sorted(zip(zero_shot_results[i]['labels'], zero_shot_results[i]['scores']), key=lambda x: x[1], reverse=True)
    for label, score in sorted_labels_scores:
        print(f"  - {label}: {score:.4f}")
    print("\n" + "-" * 50 + "\n")
```

```python
def classify_statement(text):
    text = text.lower()
    mapping = {
        "customs": ["port", "tariff", "border", "duty", "clearance"],
        "credit": ["limit", "overdue", "balance", "finance"],
        "damage": ["broken", "leaking", "cracked", "water", "dropped"],
        "quarantine": ["test", "lab", "microbial", "contamination"]
    }

    for category, keywords in mapping.items():
        if any(word in text for word in keywords):
            return category.capitalize()
    return "Other"

category_array = [classify_statement(s) for s in test_array]
```
