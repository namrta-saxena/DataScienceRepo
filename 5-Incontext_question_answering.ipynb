from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, FewShotChatMessagePromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 1. Define your examples (The "Learning" part)
examples = [
    {"input": "That meeting was a total snooze fest.", "output": "The recent assembly lacked engagement and productivity."},
    {"input": "I'm swamped right now, can't talk.", "output": "I am currently managing a high volume of tasks and am unavailable for discussion."},
    {"input": "Let's circle back on this later.", "output": "We shall revisit this topic at a more convenient juncture."},
]

# 2. Define the Example Template
example_prompt = ChatPromptTemplate.from_messages(
    [
        ("human", "{input}"),
        ("ai", "{output}"),
    ]
)

# 3. Create the Few-Shot Prompt Template
few_shot_prompt = FewShotChatMessagePromptTemplate(
    example_prompt=example_prompt,
    examples=examples,
)

# 4. Assemble the Final Prompt
final_prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a professional business translator. Convert slang into formal corporate language."),
        few_shot_prompt,
        ("human", "{input}"),
    ]
)

# 5. Build the LCEL Chain
model = ChatOpenAI(model="gpt-4o", temperature=0)
chain = final_prompt | model | StrOutputParser()

# 6. Execute
response = chain.invoke({"input": "This project is a dumpster fire."})
print(response)
